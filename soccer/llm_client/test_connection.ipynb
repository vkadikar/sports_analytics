# Gemini Connection Test

Smoke test for the Gemini API before wiring up **Commish Bot**.

Requires `GEMINI_API_KEY` (or `GOOGLE_API_KEY`) in `soccer/config/.env`, e.g.:

```
export GEMINI_API_KEY=your-key-here
```

````python
import sys
from pathlib import Path

# Make soccer/ importable so we can reuse the shared .env loader
sys.path.insert(0, str(Path.cwd().parent))
from fantrax_wrapper import load_env_file

load_env_file(str(Path.cwd().parent / "config" / ".env"))
````

````python
from google import genai
from google.genai.types import HttpOptions

client = genai.Client(http_options=HttpOptions(api_version="v1"))
response = client.models.generate_content(
    model='gemini-3.8-flash',
    contents='How does AI work?',
)
print(response.text)
````
